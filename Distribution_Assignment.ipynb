Question 6 : Explain why the uniform distribution is a good model for the outcome of rolling a fair die.

Answer 6: The uniform distribution is a good model for the outcome of rolling a die because:

A fair die has six possible outcomes (1, 2, 3, 4, 5, 6).

Each outcome has an equal probability of occurring.

Since a uniform distribution assumes that all outcomes are equally likely, it perfectly matches the behavior of a fair die.

Question 8: How does log transformation help in stabilizing variance and making data more normally distributed?

Answer 8:
Log transformation replaces each value with its logarithm to reduce skewness and stabilize variance. It compresses large values more than small ones, making the data more balanced and closer to a normal distribution. This helps improve the accuracy of statistical analysis.
